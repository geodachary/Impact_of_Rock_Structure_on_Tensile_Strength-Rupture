```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.tri as tri

# --------------------------------------------------
# Load data
# --------------------------------------------------

df = pd.read_csv(
    "selected_all_samples.csv",
    delimiter=",",
    header=0,
    index_col=0
)

# --------------------------------------------------
# Expected columns
# --------------------------------------------------

expected_columns = [
    'Rock_type',
    'Angle',
    'Diameter_mm',
    'Thickness_mm',
    'H/D_ratio',
    'Area_mm^2',
    'Volume_mm^3',
    'Bulk_Density',
    'Porosity_percent',
    'Density',
    'Load_(KN)',
    'Tensile_strength_Mpa',
    'Mass',
    'Modulus_of_Elasticity',
    'Lateral_strain',
    'Axial_strain',
    'Poisson_Ratio',
    'Shear_Modulus',
    'Radians',
    'UCS_(Mpa)',
    'Cohesion',
    'Friction_Angle'
]

# --------------------------------------------------
# Check data
# --------------------------------------------------

print("Rock types:")
print(df['Rock_type'].unique())

print("\nAngles:")
print(np.sort(df['Angle'].unique()))

print("\nOriginal dataframe shape:")
print(df.shape)

# --------------------------------------------------
# Calculate mean for each Rock_type and Angle
# --------------------------------------------------

mean_df = (
    df.groupby(['Rock_type', 'Angle'], as_index=False)
      .mean(numeric_only=True)
      .sort_values(['Rock_type', 'Angle'])
      .reset_index(drop=True)
)

# --------------------------------------------------
# Add SN column as first column
# --------------------------------------------------

mean_df.insert(
    0,
    'SN',
    np.arange(1, len(mean_df) + 1)
)

# --------------------------------------------------
# Display results
# --------------------------------------------------

print("\nMean data based on Rock Type and Angle:")
print(mean_df)

print("\nMean dataframe shape:")
print(mean_df.shape)

# --------------------------------------------------
# Separate the two rock types
# --------------------------------------------------

augen_mean = (
    mean_df[mean_df['Rock_type'] == 'Augen Gneiss']
    .reset_index(drop=True)
)

psammitic_mean = (
    mean_df[mean_df['Rock_type'] == 'Psammitic Schist']
    .reset_index(drop=True)
)

print("\nAugen Gneiss mean data:")
print(augen_mean)

print("\nPsammitic Schist mean data:")
print(psammitic_mean)

print("\nNumber of Augen Gneiss mean rows:", len(augen_mean))
print("Number of Psammitic Schist mean rows:", len(psammitic_mean))

# --------------------------------------------------
# Save results to Excel
# --------------------------------------------------

output_file = "tensile_samples_data.csv"

mean_df.to_csv(
    output_file,
    index=False
)

print(f"\nMean data successfully saved to: {output_file}")
```
